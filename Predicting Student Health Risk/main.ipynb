```python
target_column = "health_condition"
id_column = "id"

X_train = train_df.drop(columns=[target_column, id_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[id_column])
test_ids = test_df[id_column]
print(test_ids.head())
```

```python
numerical_columns = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print(numerical_columns)
print(categorical_columns)
```

```python
for numerical_column in numerical_columns:
    train_df[numerical_column] = train_df[numerical_column].fillna(
        train_df[numerical_column].median()
    )

for categorical_column in categorical_columns:
    train_df[categorical_column] = train_df[categorical_column]. fillna(
        "unknown"
    )

train_df.isnull().sum()
```
